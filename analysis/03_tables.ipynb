```r
dummy_prefixes = ['had_R_before_','active_principle','ab_group_']
df_original = pd.DataFrame(index=microbiology_icu_charlson_lab_sofa_vitals_urinedf_by_stay_orange_imputed.index)
for prefix in dummy_prefixes:
    # Identificar las columnas que empiezan con el prefijo
    columns = [col for col in microbiology_icu_charlson_lab_sofa_vitals_urinedf_by_stay_orange_imputed.columns if col.startswith(prefix)]
    # Encontrar la columna con el valor máximo por fila y eliminar el prefijo para obtener la categoría original
    df_original[prefix.rstrip('_')] = microbiology_icu_charlson_lab_sofa_vitals_urinedf_by_stay_orange_imputed[columns].idxmax(axis=1).str.replace(prefix, '')

print(df_original)
```

```r
# Eliminar las columnas dummy del DataFrame original
columns_to_drop = [col for prefix in dummy_prefixes for col in microbiology_icu_charlson_lab_sofa_vitals_urinedf_by_stay_orange_imputed.columns if col.startswith(prefix)]
df_without_dummies = microbiology_icu_charlson_lab_sofa_vitals_urinedf_by_stay_orange_imputed.drop(columns=columns_to_drop)

# Unir el DataFrame con las categorías originales al DataFrame sin las columnas dummy
df_final = pd.concat([df_without_dummies, df_original], axis=1)
```
