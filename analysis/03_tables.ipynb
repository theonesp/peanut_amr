```r
# Instalar y cargar la librería
!pip install tableone --quiet
from tableone import TableOne

# Definir la variable objetivo
target_variable = "interpretation"

# Convertir la variable objetivo a valores binarios
if df[target_variable].dtype == bool or df[target_variable].nunique() == 2:
    df[target_variable] = df[target_variable].astype(int)

# Eliminar filas con valores nulos en la variable objetivo
df = df.dropna(subset=[target_variable])

# # Variables a excluir en el modelo
# vars_excluir = ["MDRO Detected in Aggregated Blocks Before Suspicion", "block", "interpretation",
#                 "correct_preemptive_ab", "subject_id", "hadm_id", "stay_id"]

# Seleccionar solo las variables que entran en el modelo
# variables_modelo = [col for col in df.columns if col not in vars_excluir + [target_variable]]
variables_modelo = [col for col in df.columns if col not in  [target_variable]]


# Identificar variables categóricas y continuas
categorical_vars = [col for col in variables_modelo if df[col].dtype == 'object' or df[col].nunique() < 10]
continuous_vars = [col for col in variables_modelo if col not in categorical_vars]

# Manejar valores NaN en variables categóricas para evitar errores
df[categorical_vars] = df[categorical_vars].astype(str).fillna("Missing")

# Crear la tabla descriptiva con TableOne, estratificada por la variable objetivo
table1 = TableOne(df, columns=variables_modelo, categorical=categorical_vars,
                  missing=True, groupby=target_variable, pval=True)

# Mostrar la tabla
print(table1)
```
